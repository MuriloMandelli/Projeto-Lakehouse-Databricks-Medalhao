Cria os **schemas** (databases) de cada camada da arquitetura medalhão e o **volume** onde ficam os
arquivos brutos. Todos ficam no catálogo **`workspace`** do Databricks Free Edition.

| Objeto | Tipo | Conteúdo |
|---|---|---|
| `workspace.landing` | schema | Camada de pouso |
| `workspace.landing.dados` | volume | Os CSV extraídos do SQL Server |
| `workspace.bronze` | schema | Cópia fiel da landing em Delta + metadados |
| `workspace.silver` | schema | Dados limpos (Data Quality) |
| `workspace.gold` | schema | Modelo dimensional (Kimball) |

É a **1ª tarefa da Job**. Usa `IF NOT EXISTS`, então pode rodar várias vezes sem erro.

```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.landing
COMMENT 'Camada landing: arquivos brutos (CSV) extraídos do SQL Server';

CREATE VOLUME IF NOT EXISTS workspace.landing.dados
COMMENT 'Volume com os CSV extraídos do banco LojaDB';

CREATE SCHEMA IF NOT EXISTS workspace.bronze
COMMENT 'Camada bronze: cópia da landing em Delta, com metadados de ingestão';

CREATE SCHEMA IF NOT EXISTS workspace.silver
COMMENT 'Camada silver: dados limpos e padronizados (Data Quality)';

CREATE SCHEMA IF NOT EXISTS workspace.gold
COMMENT 'Camada gold: modelo dimensional (Ralph Kimball)';
```
